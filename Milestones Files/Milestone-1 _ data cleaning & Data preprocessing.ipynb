```python
# Fill with mean
df['Beds'] = df['Beds'].fillna(df['Beds'].mean())
 
# Fill with median
df['Daily_Patients'] = df['Daily_Patients'].fillna(df['Daily_Patients'].median())
 
# Fill with mode
df['City'] = df['City'].fillna(df['City'].mode()[0])
 
# Forward fill
df['Hospital_Type'] = df['Hospital_Type'].ffill()
 
# Backward fill
df['Telemedicine'] = df['Telemedicine'].bfill()
 
# Fill with a fixed value
df['Overall_Trend'] = df['Overall_Trend'].fillna("Unknown")
 
# Remaining numeric columns -> median
num_cols = ['Avg_Wait_Time_Min', 'Cleanliness_Rating', 'Food_Service_Rating',
            'Patient_Hospitality_Rating', 'CCTV_Coverage_Percentage',
            'Security_Staff', 'Infection_Control_Score', 'Patient_Satisfaction']
 
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
 
# Remaining text columns -> mode
text_cols = ['Contactless_Registration', 'Online_Appointment',
             'Digital_Health_Records', 'AI_Assisted_Services']
 
for col in text_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
 
print(df.isna().sum())
 
df.to_csv("cleaned.csv", index=False)
```

### Handling Outliers

```python
def beds_category(beds):
    if beds < 300:
        return "Small"
    elif beds < 800:
        return "Medium"
    elif beds < 1200:
        return "Large"
    else:
        return "Very Large"
df["Beds_1"] = df["Beds"].apply(beds_category)
```

```python
def wait_time_category(minutes):
    if minutes < 30:
        return "Low"
    elif minutes < 75:
        return "Medium"
    elif minutes < 120:
        return "High"
    else:
        return "Very High"
df["Avg_Wait_Time_Min_1"] = df["Avg_Wait_Time_Min"].apply(wait_time_category)
```

```python
def satisfaction_category(score):
    if score < 2.5:
        return "Poor"
    elif score < 3.5:
        return "Average"
    elif score < 4.5:
        return "Good"
    else:
        return "Excellent"
df["Patient_Satisfaction_1"] = df["Patient_Satisfaction"].apply(satisfaction_category)
```

```python
def cctv_category(pct):
    if pct < 50:
        return "Low"
    elif pct < 75:
        return "Medium"
    else:
        return "High"
df["CCTV_Coverage_Percentage_1"] = df["CCTV_Coverage_Percentage"].apply(cctv_category)
```

### ENCODING CATEGORICAL VARIABLES

```python
# One-Hot Encoding for multi-category columns
df = pd.get_dummies(
    df,
    columns=["City", "Hospital_Type", "Overall_Trend"],
    dtype=int
)
```

### Features Engineering

```python
# Patients handled per bed - workload/capacity indicator
df["Patients_per_Bed"] = df["Daily_Patients"] / df["Beds"]
 
```

```python
# Security staff available per 100 beds - security adequacy indicator
df["Security_per_100_Beds"] = (df["Security_Staff"] / df["Beds"]) * 100
```

### Features Selection

```python
# Hospital_ID and Hospital_Name are identifiers, not useful for analysis/modeling
df = df.drop(["Hospital_ID", "Hospital_Name"], axis=1)
```
